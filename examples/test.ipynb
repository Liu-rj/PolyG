```python
# iterate over the items and print the first item that has a non-empty neighboring brand field
example_key = list(data.keys())[0]
for key in data[example_key].keys():
    if data[example_key][key]['neighbors']['brand'] != []:
        print(key, data[example_key][key])
        break
```

```python
# construct a networkx graph using this json file
import networkx as nx

G = nx.Graph()

# add nodes
for node_type in data.keys():
    for key, value in data[node_type].items():
        if node_type == "paper_nodes":
            data[node_type][key]["features"]["label"] = ", ".join(data[node_type][key]["features"]["label"])
        G.add_node(key, **data[node_type][key]["features"])

# add edges
for node_type in data.keys():
    for key, value in data[node_type].items():
        for relation, neighbors in data[node_type][key]["neighbors"].items():
            if isinstance(neighbors, list):
                for edge in neighbors:
                    G.add_edge(key, edge, relation=relation)
            else:
                G.add_edge(key, edge, relation=relation)
```
